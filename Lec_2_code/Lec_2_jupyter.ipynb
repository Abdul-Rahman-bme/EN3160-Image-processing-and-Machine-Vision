```python
import cv2 as cv
import matplotlib.pyplot as plt

im = cv.imread('D:/Abdul Rahman/Engineering_UOM/.Sem 5/EN3160 - Image Processing and Computer Vision/Lec_2_vscode/emma.jpg')
assert im is not None, "file could not be read, check with os.path.exists()"

im_green=im.copy()
im_green[:,:,0]=0 # set green channel to 0
im_green[:,:,2]=0 # set red channel to 0

im_red=im.copy()
im_red[:,:,0]=0 # set blue channel to 0
im_red[:,:,1]=0 # set green channel to 0

im_blue=im.copy()
im_blue[:,:,1]=0 # set green channel to 0   
im_blue[:,:,2]=0 # set red channel to 0

fig, ax = plt.subplots(nrows=2,ncols=2, figsize=(6,4))
ax[0,0].imshow(cv.cvtColor(im_green, cv.COLOR_BGR2RGB))
ax[0,0].set_title('Green Channel')
ax[0,0].axis('off')

ax[0,1].imshow(cv.cvtColor(im_red, cv.COLOR_BGR2RGB))
ax[0,1].set_title('Red Channel')
ax[0,1].axis('off')

ax[1,0].imshow(cv.cvtColor(im_blue, cv.COLOR_BGR2RGB))
ax[1,0].set_title('Blue Channel')
ax[1,0].axis('off')

ax[1,1].imshow(cv.cvtColor(im, cv.COLOR_BGR2RGB))
ax[1,1].set_title('Original Image')
ax[1,1].axis('off')
plt.show()
```

```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np

f = cv.imread('D:/Abdul Rahman/Engineering_UOM/.Sem 5/EN3160 - Image Processing and Computer Vision/Lec_2_vscode/emma.jpg', cv.IMREAD_GRAYSCALE)
assert f is not None, "file could not be read, check with os.path.exists()"

t = np.arange(256, dtype=np.uint8)  # Identity transformation

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(12, 4))
ax.plot(t, color='black')
ax.set_xlim(0, 255)
ax.set_ylim(0, 255)
ax.set_xlabel('Input Intensity')
ax.set_ylabel('Output Intensity')
ax.set_title('Identity Transformation')
plt.show()

transformed = cv.LUT(f, t)  # Apply the LUT to the actual image

fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(12, 4))

ax[0].imshow(f, cmap='gray', vmin=0, vmax=255)
ax[0].set_title('Original Image')
ax[0].axis('off')

ax[1].imshow(transformed, cmap='gray', vmin=0, vmax=255)
ax[1].set_title('Transformed Image')
ax[1].axis('off')

plt.show()
```

```python
from pathlib import Path

import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np


def inspect_image(path: str) -> None:
    """Load an image, display it correctly, and print its properties."""
    image_path = Path(path)

    if not image_path.exists():
        raise FileNotFoundError(f"Image not found: {image_path}")

    bgr = cv.imread(str(image_path), cv.IMREAD_COLOR)
    if bgr is None:
        raise ValueError(f"OpenCV could not decode: {image_path}")

    rgb = cv.cvtColor(bgr, cv.COLOR_BGR2RGB)

    print(f"Shape: {bgr.shape}")
    print(f"Data type: {bgr.dtype}")
    print(f"Number of array elements: {bgr.size}")
    print(f"Minimum value: {bgr.min()}")
    print(f"Maximum value: {bgr.max()}")
    print(f"Mean BGR values: {bgr.mean(axis=(0, 1))}")

    plt.figure(figsize=(8, 6))
    plt.imshow(rgb)
    plt.title("RGB display")
    plt.axis("off")
    plt.show()


inspect_image("D:/Abdul Rahman/Engineering_UOM/.Sem 5/EN3160 - Image Processing and Computer Vision/Lec_2_vscode/emma.jpg")
```
