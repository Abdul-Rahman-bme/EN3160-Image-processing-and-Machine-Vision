```python
from pathlib import Path

import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np


def inspect_image(path: str) -> None:
    """Load an image, display it correctly, and print its properties."""
    image_path = Path(path)

    if not image_path.exists():
        raise FileNotFoundError(f"Image not found: {image_path}")

    bgr = cv.imread(str(image_path), cv.IMREAD_COLOR)
    if bgr is None:
        raise ValueError(f"OpenCV could not decode: {image_path}")

    rgb = cv.cvtColor(bgr, cv.COLOR_BGR2RGB)

    print(f"Shape: {bgr.shape}")
    print(f"Data type: {bgr.dtype}")
    print(f"Number of array elements: {bgr.size}")
    print(f"Minimum value: {bgr.min()}")
    print(f"Maximum value: {bgr.max()}")
    print(f"Mean BGR values: {bgr.mean(axis=(0, 1))}")

    plt.figure(figsize=(8, 6))
    plt.imshow(rgb)
    plt.title("RGB display")
    plt.axis("off")
    plt.show()


inspect_image("D:/Abdul Rahman/Engineering_UOM/.Sem 5/EN3160 - Image Processing and Computer Vision/Lec_2_vscode/emma.jpg")
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

f = cv.imread(r"D:\Abdul Rahman\Engineering_UOM\.Sem 5\EN3160 - Image Processing and Computer Vision\EN3160-Image-processing-and-Machine-Vision\images\Color_ful_pic.jpg")

# build the windowing LUT
c = np.array([(80, 20), (180, 220)])
t1 = np.linspace(0, c[0,1], c[0,0] + 1 - 0).astype('uint8')
t2 = np.linspace(c[0,1] + 1, c[1,1], c[1,0] - c[0,0]).astype('uint8')
t3 = np.linspace(c[1,1] + 1, 255, 255 - c[1,0]).astype('uint8')
t = np.concatenate((t1, t2, t3), axis=0).astype('uint8')

g = t[f]   # exact same indexing trick as your negative transform — works fine on the 3-channel array

fig, ax = plt.subplots(1, 3, figsize=(15, 8))
ax[0].imshow(cv.cvtColor(f, cv.COLOR_BGR2RGB))
ax[0].set_title('Original Image')
ax[1].imshow(cv.cvtColor(g, cv.COLOR_BGR2RGB))
ax[1].set_title('Windowed Image')
ax[2].plot(t)
ax[2].set_title('Transform')
for a in ax[:2]:
    a.axis('off')
plt.show()
```

# Gamma Correction

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

f = cv.imread("D:\Abdul Rahman\Engineering_UOM\.Sem 5\EN3160 - Image Processing and Computer Vision\EN3160-Image-processing-and-Machine-Vision\images\window.jpg")

def gamma_lut(gamma):
    return np.array([(i/255.0)**gamma * 255 for i in np.arange(256)]).astype(np.uint8)

gammas = [0.4, 1.0, 2.5]   # <1 brightens, =1 identity, >1 darkens
results = [cv.LUT(f, gamma_lut(g)) for g in gammas]

fig, ax = plt.subplots(1, len(gammas), figsize=(15, 8))
for a, g, gamma in zip(ax, results, gammas):
    a.imshow(cv.cvtColor(g, cv.COLOR_BGR2RGB))
    a.set_title(f'γ = {gamma}')
    a.axis('off')
plt.show()
```

# Histogram Equilization

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

f_color = cv.imread("D:\Abdul Rahman\Engineering_UOM\.Sem 5\EN3160 - Image Processing and Computer Vision\EN3160-Image-processing-and-Machine-Vision\images\emma.jpg")
f = cv.cvtColor(f_color, cv.COLOR_BGR2GRAY)   # start with grayscale, the textbook case

g_builtin = cv.equalizeHist(f)

hist, _ = np.histogram(f.flatten(), bins=256, range=(0, 256))
cdf = hist.cumsum()
M, N = f.shape
L = 256
t = np.array([(L - 1) / (M * N) * cdf[k] for k in range(256)]).astype(np.uint8)
g_manual = t[f]

fig, ax = plt.subplots(2, 2, figsize=(12, 10))

ax[0,0].imshow(f, cmap='gray', vmin=0, vmax=255)
ax[0,0].set_title('Original'); ax[0,0].axis('off')

ax[0,1].imshow(g_manual, cmap='gray', vmin=0, vmax=255)
ax[0,1].set_title('Equalized'); ax[0,1].axis('off')

ax[1,0].hist(f.flatten(), bins=256, range=(0,255), color='gray')
ax[1,0].set_title('Original histogram')

ax[1,1].hist(g_manual.flatten(), bins=256, range=(0,255), color='gray')
ax[1,1].set_title('Equalized histogram')

plt.show()

# confirm your manual version matches OpenCV's
print('Max pixel difference vs cv.equalizeHist:', np.abs(g_manual.astype(int) - g_builtin.astype(int)).max())
```
